```python

import pandas as pd
import numpy as np

# Generate 240 days of sample retail sales data
np.random.seed(42)

dates = pd.date_range("2026-01-01", periods=240, freq="D")
trend = np.arange(240) * 35
noise = np.random.normal(0, 180, 240)

sales = 4500 + trend + noise
sales = np.maximum(sales, 500).round(2)

df = pd.DataFrame({
    "date": dates,
    "items_sold": np.round(sales / 100).astype(int),
    "transactions": np.round(sales / 150).astype(int),
    "daily_sales": sales
})

df.to_csv("daily_retail_sales.csv", index=False)

print("CSV file created!")
print("Total rows:", len(df))
display(df.head())

# Download the CSV file to your computer
from google.colab import files
files.download("daily_retail_sales.csv")
```
